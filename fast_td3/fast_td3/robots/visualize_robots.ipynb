## H1 Robot Visualization

The H1 robot has 19 joints with the following structure:
- **Legs**: 5 joints per leg (hip yaw/roll/pitch, knee, ankle)
- **Arms**: 4 joints per arm (shoulder pitch/roll/yaw, elbow)
- **Torso**: 1 joint

```python
# Create H1 graph builder
gb_h1 = GraphBuilder(env_name="h1-run-v0", batch_size=1, device="cpu", robot="h1")

print(f"H1 Robot Statistics:")
print(f"  Number of joints: {gb_h1.robot.num_joints}")
print(f"  Number of edges: {gb_h1.robot.num_edges}")
```

```python
# Visualize H1
print("\nH1 Robot Graph:")
gb_h1.visualize_graph(save_path="h1_robot_graph.png")
```

- **Legs**: 6 joints per leg (hip pitch/roll/yaw, knee, ankle pitch/roll)

- **Torso**: 1 joint

```python
# Create G1 graph builder
gb_g1 = GraphBuilder(env_name="g1-run-v0", batch_size=1, device="cpu", robot="g1")

print(f"G1 Robot Statistics:")
print(f"  Number of joints: {gb_g1.robot.num_joints}")
print(f"  Number of edges: {gb_g1.robot.num_edges}")
```
